## Adding column


```python
import pandas as pd
```

```python
data = {
    "name": ["Noor", "Rahul", "Priya", "Aman", "Sneha", "Arjun", "Neha", "Vikash", "Anjali", "Rohan"],
    "age": [24, 27, 25, 30, 23, 28, 26, 31, 24, 29],
    "salary": [45000, 60000, 52000, 75000, 48000, 68000, 55000, 80000, 50000, 65000],
    "performance_score": [85, 92, 78, 95, 88, 90, 82, 97, 76, 89]
}

```

```python
data = {
    "name": ["Noor", "Rahul", None, "Aman", "Sneha", "Arjun", "Neha", "Vikash", "Anjali", "Rohan"],
    "age": [24, 27, None, 30, 23, 28, 26, 31, 24, 29],
    "salary": [45000, 60000, None, 75000, 48000, 68000, 55000, 80000, 50000, 65000],
    "performance_score": [85, 92, None, 95, 88, 90, 82, 97, 76, 89]
}
```

```python
# number of missing values in each column
print(df.isnull().sum())
```

### Removing the missing value


## dropna()
### df.dropna(axis = 0, inplace=True) if axis is 0 delete (row) 1 (col)

```python
df.dropna(inplace=True)
print(df)
```

```python
df.dropna(axis=0, subset=["name"], inplace=True)
print(df)
```

## Interpalatin
### pushing or adding estimating value
#### 1-> preserve data integarity
#### 1-> smoth trends
#### 1-> avoid data los

```python
data = {
    "name": ["Noor", "Rahul", "Priya", "Aman", "Sneha", "Arjun", "Neha", "Vikash", "Anjali", "Rohan"],
    "age": [24, None, 25, 30, 23, 28, 26, 31, 24, 29],
    "salary": [45000, 60000, 52000, 75000, 48000, 68000, 55000, 80000, 50000, 65000],
    "performance_score": [85, 92, 78, 95, 88, 90, 82, 97, 76, 89]
}

df = pd.DataFrame(data)
print(df) # before adding the missing values
```

```python
# df.interpolate() method is used to fill the missing values in a DataFrame by using interpolation techniques. Interpolation is a method of estimating missing values based on the values of neighboring data points.
df['age'] = df['age'].interpolate(method='linear', inplace=True, axis=0) # axis=0 means we are filling the missing values in rows
print(df) # after adding the missing values
```

## Grouping 

```python
data = {
    "name": ["wasim bhai", "Rahul", "Priya", "Aman", "Sneha", "Arjun", "Neha", "Vikash", "Anjali", "Rohan"],
    "age": [24, None, 25, 30, 23, 28, 26, 31, 24, 29],
    "salary": [45000, 60000, 52000, 75000, 48000, 68000, 55000, 80000, 50000, 65000],
    "performance_score": [85, 92, 78, 95, 88, 90, 82, 97, 76, 89]
}
df = pd.DataFrame(data)
"""
df.groupby('age') this will create a group of data based on the age column. It will return a DataFrameGroupBy object, which can be used to perform various operations on the grouped data.
age = 12 [2223, 5543]
age = 14 [24, 25]

"""
group = df.groupby('salary')['salary'].sum()
print(group)
```

## marging and joining

## pd.mearge(name_of_datafram1, name_of_datafram2, on="similar_coulumn_name", how= "type)


```python
df_merge1 = pd.DataFrame({
    "CustomerID": [1, 2, 3, 4, 5],
    "CustomerName": ["John", "Alice", "Bob", "Eve", "Charlie"]
})

df_merge2 = pd.DataFrame({
    "CustomerID": [1, 2, 3, 4, 5, 6],
    "OrderID": [101, 102, 103, 104, 105, 106]
})

df_merged = pd.merge(df_merge1, df_merge2, on="CustomerID", how="inner") # inner join# 
# how = "inner" will return only the rows that have matching values in both DataFrames.
# how = "outer" will return all rows from both DataFrames, with NaN in places where there is no match.
# how = "left" will return all rows from the left DataFrame, and the matched rows from the right DataFrame. If there is no match, the result will contain NaN for the right DataFrame's columns.
# how = "right" will return all rows from the right DataFrame, and the matched rows from the left DataFrame. If there is no match, the result will contain NaN for the left DataFrame's columns.


print(df_merged)
```

## concatination

```python
df_regin1 = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "Region": ["North", "South", "East", "West"],
})

df_regin2 = pd.DataFrame({
    "customer_id": [5, 6, 7, 8],
    "Region": ["North", "South", "East", "West"],
})

df_concat = pd.concat([df_regin1, df_regin2], axis=0, ignore_index=True) # axis=0 means we are concatenating the rows
print(df_concat)

```
