```python
data = {
    "name": ["Noor", "Rahul", "Priya", "Aman", "Sneha", "Arjun", "Neha", "Vikash", "Anjali", "Rohan"],
    "age": [24, 27, 25, 30, 23, 28, 26, 31, 24, 29],
    "salary": [45000, 60000, 52000, 75000, 48000, 68000, 55000, 80000, 50000, 65000],
    "performance_score": [85, 92, 78, 95, 88, 90, 82, 97, 76, 89]
}

#print(pd.DataFrame(data))
df = pd.DataFrame(data)

# Accessing specific columns
name_column = df['name']
print("Name Column:", name_column)

# Accessing multiple columns
selected_columns = df[['name', 'age']]
print("Selected Columns:\n", selected_columns)

# Filtering rows based on a condition
filtered_df = df[df['age'] > 24]
print(filtered_df)
```

```python
data = {
    "name": ["wasim", "Noor", "Priya", "Aman", "Sneha", "Arjun", "Neha", "Vikash", "Anjali", "Rohan"],
    "age": [24, 27, 25, 30, 23, 28, 26, 31, 24, 29],
    "salary": [45000, 60000, 52000, 75000, 48000, 68000, 55000, 80000, 50000, 65000],
    "performance_score": [85, 92, 78, 95, 88, 90, 82, 97, 76, 89]
}

df = pd.DataFrame(data)

# Accessing specific row
high_salary_row = df[df['salary'] > 60000]
print("Rows with Salary greater than 60000:\n", high_salary_row)

# Accessing specific row with two conditions
high_salary_and_age_row = df[(df['salary'] > 60000) & (df['age'] < 30)]
print("Rows with Salary greater than 60000 and Age less than 30:\n", high_salary_and_age_row)

print("________________________________________________________")
# Accessing specific row with two conditions
high_salary_or_age_row = df[(df['salary'] > 60000) | (df['age'] < 30)]
print("Rows with Salary greater than 60000 or Age less than 30:\n", high_salary_or_age_row)

```
